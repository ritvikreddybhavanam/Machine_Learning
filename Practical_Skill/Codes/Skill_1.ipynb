```python
num_cols = df.select_dtypes(include="number").columns

fig, axes = plt.subplots(3, 3, figsize=(15, 12))

for i, column in enumerate(num_cols):
    row, colpos = divmod(i, 3)

    sns.histplot(df[column].dropna(), bins=30, kde=True, ax=axes[row, colpos])

    axes[row, colpos].set_title(f'Distribution of {column}')
    axes[row, colpos].set_xlabel(column)
    axes[row, colpos].set_ylabel('Frequency')

for i in range(len(num_cols), 9):
    row, colpos = divmod(i, 3)
    axes[row, colpos].set_visible(False)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x="Sex", hue="Survived")

plt.title("Survival Analysis by Sex")
plt.xlabel("Sex")
plt.ylabel("Number of Passangers")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x="Pclass", hue="Survived")

plt.title("Survival Analysis by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passangers")
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(data=df, x="Age", hue="Survived", bins=30, kde=True)

plt.title("Survival Analysis by Age")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
plt.figure(figsize=(10, 7))

numeric_df = df.select_dtypes(include="number")

sns.heatmap(numeric_df.corr(), annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Correlation Heatmap")
plt.show()
```
